## Step 4 — Create the target column `is_attack`
Add a new column that is `0` when the connection is benign and `1` when it's an attack.

**Hints:**
- The `label` column contains the string `"normal"` for benign traffic, and various attack names otherwise.
- The comparison `df["label"] != "normal"` produces a column of `True`/`False`.
- Chaining `.astype(int)` turns `True`→`1` and `False`→`0`.
- Assign the result to `df["is_attack"]`.

```python
# TODO: df["is_attack"] = ...
df["is_attack"] = (df["label"]!="normal").astype(int)
```
